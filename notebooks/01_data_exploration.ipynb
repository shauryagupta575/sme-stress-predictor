```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

pd.set_option('display.max_columns', 50)
pd.set_option('display.max_rows', 100)

print("✅ All libraries imported successfully")
```

```python
# Dataset 1 - Commodity prices (agmarknet)
df_commodity = pd.read_csv('../data/raw/agmarknet/commodity_prices.csv')

print("=== COMMODITY PRICES ===")
print(f"Shape: {df_commodity.shape}")
print(f"Columns: {list(df_commodity.columns)}")
print(f"Date range: {df_commodity.iloc[0,0]} to {df_commodity.iloc[-1,0]}")
print(f"Missing values:\n{df_commodity.isnull().sum()}")
print()
print(df_commodity.head(5))
```

```python
# Dataset 2 - RBI/Credit data
df_cibil = pd.read_excel('../data/raw/rbi/External_Cibil_Dataset.xlsx')
df_bank = pd.read_excel('../data/raw/rbi/Internal_Bank_Dataset.xlsx')
df_credit = pd.read_csv('../data/raw/lending_club/credit_risk.csv')

print("=== EXTERNAL CIBIL DATASET ===")
print(f"Shape: {df_cibil.shape}")
print(f"Columns: {list(df_cibil.columns)}")
print(f"Missing values:\n{df_cibil.isnull().sum()}")
print()

print("=== INTERNAL BANK DATASET ===")
print(f"Shape: {df_bank.shape}")
print(f"Columns: {list(df_bank.columns)}")
print()

print("=== CREDIT RISK DATASET ===")
print(f"Shape: {df_credit.shape}")
print(f"Columns: {list(df_credit.columns)}")
print(df_credit.head(3))
```

```python
# Dataset 3 - GST data
df_gst_X_train = pd.read_csv('../data/raw/gst/X_Train_Data_Input.csv')
df_gst_y_train = pd.read_csv('../data/raw/gst/Y_Train_Data_Target.csv')
df_gst_X_test  = pd.read_csv('../data/raw/gst/X_Test_Data_Input.csv')
df_gst_y_test  = pd.read_csv('../data/raw/gst/Y_Test_Data_Target.csv')
df_goods       = pd.read_csv('../data/raw/gst/Goods.csv', encoding='latin-1')  # ← fix here

print("=== GST PREDICTIVE DATASET ===")
print(f"Train X shape: {df_gst_X_train.shape}")
print(f"Train y shape: {df_gst_y_train.shape}")
print(f"Test X shape:  {df_gst_X_test.shape}")
print(f"Columns: {list(df_gst_X_train.columns)}")
print()
print(df_gst_X_train.head(3))
print()

print("=== GOODS GST RATES ===")
print(f"Shape: {df_goods.shape}")
print(df_goods.head(3))
```

```python
print("=" * 50)
print("DATASET SUMMARY")
print("=" * 50)

datasets = {
    "Commodity Prices": df_commodity,
    "CIBIL External": df_cibil,
    "Bank Internal": df_bank,
    "Credit Risk": df_credit,
    "GST Train X": df_gst_X_train,
    "GST Train y": df_gst_y_train,
    "Goods GST": df_goods
}

for name, df in datasets.items():
    print(f"{name:25} | Rows: {df.shape[0]:8,} | Cols: {df.shape[1]:4} | Nulls: {df.isnull().sum().sum()}")
```
